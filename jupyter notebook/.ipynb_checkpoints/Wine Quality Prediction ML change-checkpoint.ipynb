### Create Quality Categories
MODIFIED: 3 balanced categories
- ≤ 5 → Average
- 6-7 → Good
- ≥ 8 → Excellent

```python
def quality_category(score):
    if score <= 5:
        return "Average"
    elif score <= 7:
        return "Good"
    else:
        return "Excellent"

df['quality_cat'] = df['quality'].apply(quality_category)
```

```python
df['quality_cat'].value_counts()
```

```python
df['quality'].value_counts()
```

```python
plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
sns.countplot(x='quality_cat', data=df, order=['Average', 'Good', 'Excellent'])
plt.title("Wine Quality Category Distribution")
plt.xticks(rotation=30)

plt.subplot(1, 2, 2)
sns.countplot(x='quality', data=df)
plt.title("Wine Quality Score Distribution")

plt.tight_layout()
plt.show()
```

- Key Insights 
- Alcohol shows a strong positive correlation with wine quality.
- Volatile acidity has a strong negative impact on wine quality.
- Most wines fall into the average quality range (scores 5–6), indicating class imbalance.

## MODEL BUILDING

##### feature and target selection

```python
x = df.drop(columns=['quality', 'quality_cat','good'])
y = df['quality_cat']
```

#### Train Test Split

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(x, y,test_size=0.25,random_state=42,stratify=y)
```

- shape of train and test

```python
X_train.shape
```

```python
X_test.shape
```

```python
print("\nTraining set distribution:")
print(y_train.value_counts())
print("\nTest set distribution:")
print(y_test.value_counts())
```

## applying smote on training data only

```python
from imblearn.over_sampling import SMOTE

smote = SMOTE(random_state=42, k_neighbors=5)
X_train_sm, y_train_sm = smote.fit_resample(X_train, y_train)
```

```python
print("Before SMOTE:")
print(y_train.value_counts())

print("\nAfter SMOTE:")
print(y_train_sm.value_counts())
```

#### Feature Scaling

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
X_train_sm_scaled = scaler.fit_transform(X_train_sm)
X_test_scaled = scaler.transform(X_test)
```

#### Handle Class Imbalance with RandomOverSampler
**Using RandomOverSampler instead of SMOTE to avoid k_neighbors issue**

### training Model

```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report,confusion_matrix
```

- KNN classifier

```python
knn = KNeighborsClassifier(n_neighbors=7)
knn.fit(X_train_sm_scaled, y_train_sm)
```

- Logistic Regression 

```python
lr = LogisticRegression(max_iter=1000)
lr.fit(X_train_sm_scaled, y_train_sm)
```

- Support Vector Classifier

```python
svm = SVC(kernel='rbf')
svm.fit(X_train_sm_scaled, y_train_sm)
```

- Decision Tree Classifier

```python
dt = DecisionTreeClassifier(random_state=42)
dt.fit(X_train_sm_scaled, y_train_sm)
```

- Random Forest Classifier

```python
#rf = RandomForestClassifier(n_estimators=200,max_depth=20,random_state=42)
rf = RandomForestClassifier(n_estimators=200,random_state=42)
rf.fit(X_train_sm_scaled, y_train_sm)
```

### Evaluation matrix

- KNN model

```python
y_pred_knn = knn.predict(X_test)

print("KNN Accuracy:", accuracy_score(y_test, y_pred_knn))
print(confusion_matrix(y_test, y_pred_knn))
print(classification_report(y_test, y_pred_knn))
```

- Logistic regression model

```python
lr_pred = lr.predict(X_test)
print("Logistic Regression Accuracy:", accuracy_score(y_test, lr_pred))
print(confusion_matrix(y_test, lr_pred))
print(classification_report(y_test,lr_pred))
```

- SVC

```python
svm_pred = svm.predict(X_test)
print("SVC Accuracy:", accuracy_score(y_test, svm_pred))
print(confusion_matrix(y_test, svm_pred))
print(classification_report(y_test,svm_pred))
```

- Decission Tree model

```python
dt_pred = dt.predict(X_test)
print("Decission Tree Accuracy:", accuracy_score(y_test, dt_pred))
print(confusion_matrix(y_test, dt_pred))
print(classification_report(y_test,dt_pred))
```

- Random Forest model

```python
rf_pred = rf.predict(X_test)
print("Random Forest Accuracy:", accuracy_score(y_test, rf_pred))
print(confusion_matrix(y_test, rf_pred))
print(classification_report(y_test,rf_pred))
```

```python
import pickle

# Save trained model
with open("wine_quality_model.pkl", "wb") as f:
    pickle.dump(rf, f)

# Save scaler
with open("scaler.pkl", "wb") as f:
    pickle.dump(scaler, f)

print("Model and scaler saved successfully!")
```
